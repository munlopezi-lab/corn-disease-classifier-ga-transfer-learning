### 1. Importación de TensorFlow y Dependencias
Carga de TensorFlow y módulos de Keras necesarios para redes neuronales y transferencia de aprendizaje.

```python
import tensorflow as tf

from keras.applications import EfficientNetB0
from keras.applications.efficientnet import preprocess_input
import numpy as np
```

### 2. Carga del Modelo Base EfficientNetB0
Inicialización de EfficientNetB0 preentrenado en ImageNet sin la capa de clasificación (include_top=False) para la extracción de características.

```python
base_model = EfficientNetB0(weights='imagenet', include_top=False, 
                            pooling='avg', input_shape=(224, 224, 3))
```

### 3. Verificación de Directorios
Comprobación de las rutas de datos (entrenamiento, validación, prueba) antes de iniciar la extracción.

```python
print(f"--- Procesando con EfficientNetB0: {'Maiz_data/val'} ---")
directory = 'Maiz_data/val'
dataset = tf.keras.utils.image_dataset_from_directory(
    directory,       
    image_size=(224, 224),
    batch_size=32,
    label_mode='int',
     shuffle=False
)

print(dataset.class_names)
```

### 4. Función de Extracción de Características
Definición y ejecución de la función para procesar lotes de imágenes a través de EfficientNetB0 y almacenar los tensores resultantes en archivos .npy.

```python
def extract_features(directory):
    print(f"--- Procesando con EfficientNetB0: {directory} ---")
    dataset = tf.keras.utils.image_dataset_from_directory(
        directory,
        image_size=(224, 224),
        batch_size=32,
        label_mode='int',
        shuffle=False
    )

    features = []
    labels = []

    for images, batch_labels in dataset:
        # Importante: usamos el preprocess_input específico de EfficientNet
        preprocessed_imgs = preprocess_input(images)
        batch_features = base_model.predict(preprocessed_imgs, verbose=1)
        features.append(batch_features)
        labels.append(batch_labels.numpy())
        
    return np.concatenate(features), np.concatenate(labels)



# %% 3. Ejecutar extracción
# (Asegúrate de que las carpetas existan)
X_train, y_train = extract_features('Maiz_data/train')
X_val, y_val = extract_features('Maiz_data/val')
X_test, y_test = extract_features('Maiz_data/test')

# %% 4. Guardar con nombres nuevos para no sobrescribir tus archivos de MobileNet
np.save('X_train_effnet.npy', X_train)
np.save('y_train_effnet.npy', y_train)
np.save('X_val_effnet.npy', X_val)
np.save('y_val_effnet.npy', y_val)
np.save('X_test_effnet.npy', X_test)
np.save('y_test_effnet.npy', y_test)

print("¡Extracción completada con EfficientNetB0!")
```
