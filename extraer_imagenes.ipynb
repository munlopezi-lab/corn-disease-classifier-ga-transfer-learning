### 1. Importación de Librerías
Importación de módulos necesarios para la manipulación de rutas y sistemas de archivos.

```python
import os
import shutil
import random
```

### 2. Configuración de Directorios y Clases
Definición del directorio de origen (donde se descargó o guardó el dataset PlantVillage) y estructuración de los conjuntos de datos. También se define el nombre de la carpeta donde se guradará el dataset estructurado.

```python
origen = 'PlantVillage'
destino_final = 'Maiz_data'
clases = ['Corn_(maize)___Cercospora_leaf_spot Gray_leaf_spot', 
          'Corn_(maize)___Common_rust_',
          'Corn_(maize)___Northern_Leaf_Blight',
          'Corn_(maize)___healthy']
```

### 3. Organización de Datos por Particiones
Iteración sobre los conjuntos de entrenamiento, validación y prueba para organizar las imágenes correspondientes.

```python
for split in ['train', 'val', 'test']:
    for c in clases:
        os.makedirs(os.path.join(destino_final, split, c), exist_ok=True)
```

### 4. Procesamiento y Clasificación de Imágenes
Recorrido de las clases de cultivo para procesar, copiar o estructurar las imágenes en el pipeline de datos.

```python
for c in clases:    # Recorre ambas clases
    ruta_clase = os.path.join(origen, c) # Va a la carpeta de la clase c con todas las imágenes
    fotos = os.listdir(ruta_clase)       # Toma todos los nombres de las imágenes
    random.shuffle(fotos)                # Mezcla los nombres de las imágenes para que no estén ordenados


    total_fotos = len(fotos)
    limite_train = int(total_fotos * 0.70)
    limite_val = int(total_fotos * 0.85)

    # Hace los cortes
    train_f = fotos[:limite_train]         # 700 para entrenamiento
    val_f = fotos[limite_train:limite_val] # 150 para validación
    test_f = fotos[limite_val:]            # 150 para prueba



    # Hace copias de las imágenes a donde corresponde
    for f in train_f:
        shutil.copy(os.path.join(ruta_clase, f), os.path.join(destino_final, 'train', c, f))
    for f in val_f:
        shutil.copy(os.path.join(ruta_clase, f), os.path.join(destino_final, 'val', c, f))
    for f in test_f:
        shutil.copy(os.path.join(ruta_clase, f), os.path.join(destino_final, 'test', c, f))
```
